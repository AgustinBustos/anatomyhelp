```python
from bs4 import BeautifulSoup
import pandas as pd
with open('./page.html', 'r', encoding="utf-8") as file:
    html_content = file.read()

# html_content
soup = BeautifulSoup(html_content)  #.encode('utf-8')

print(soup.prettify())
```

```python
li_elements = [dict([('link',i.find('a')['href'])]+[(j['class'][0],j.text) for j in i.find_all('p')]) for i in soup.find_all('li')]  #
df=pd.DataFrame(li_elements)
df.to_csv('./data.csv')
```
